```python
import ast

def decode_detection(payload: bytes) -> dict:
    text = payload.decode("utf-8").strip()
    if not text:
        return {"labels": [], "scores": [], "boxes": []}
    labels, scores, boxes = [], [], []
    for line in text.split("\n"):
        parts = line.split("|", 2)
        if len(parts) != 3:
            raise ValueError(f"Malformed detection line: {line}")
        label, score_str, box_str = parts
        labels.append(label)
        scores.append(float(score_str))
        boxes.append(ast.literal_eval(box_str))
    return {"labels": labels, "scores": scores, "boxes": boxes}
```

```python
from torch.utils.data import Dataset, DataLoader, IterableDataset
```

```python
import webdataset as wd
from PIL import Image
import io
class WebDatasetWrapper(IterableDataset):
    def __init__(self, tar_path, decode_fn, shuffle=False):
        dataset = wd.WebDataset(tar_path, handler=wd.handlers.warn_and_continue)
        self.dataset = dataset.shuffle(1000) if shuffle else dataset
        self.decode_fn = decode_fn
        if decode_fn is None:
            raise ValueError("decode_fn must be provided")
    
    def __iter__(self):
        for sample in self.dataset:
            image = Image.open(io.BytesIO(sample.get("img")))
            det = self.decode_fn(sample.get("det.pb"))
            text = sample.get("txt")
            yield {"image": image, "det": det, "text": text}
```

```python
wrapper = WebDatasetWrapper("/data1/zhangyang/Database/yfcc15m/shard-000001.tar", decode_detection, shuffle=True)
```

```python
import pytorch_lightning as pl
from torch.utils.data import default_collate
class WebDatasetDataModule(pl.LightningDataModule):
    def __init__(self, train_tar, val_tar, batch_size, num_workers=4, collate_fn = default_collate):
        super().__init__()
        self.train_tar = train_tar
        self.val_tar = val_tar
        self.batch_size = batch_size
        self.num_workers = num_workers
        self.collate_fn = collate_fn
    
    def train_dataloader(self):
        dataset = WebDatasetWrapper(self.train_tar, decode_detection, shuffle=True)
        return DataLoader(dataset, batch_size=self.batch_size, num_workers=self.num_workers, collate_fn=self.collate_fn)
    
    def val_dataloader(self):
        if self.val_tar is None:
            return None
        dataset = WebDatasetWrapper(self.val_tar, decode_detection, shuffle=False)
        return DataLoader(dataset, batch_size=self.batch_size, num_workers=self.num_workers, collate_fn=self.collate_fn)
```

```python
def vae_preprocess(images):
    from diffusers.image_processor import VaeImageProcessor
    processor = VaeImageProcessor(
        do_binarize=False,
        do_convert_grayscale=False,
        do_convert_rgb=False,
        do_normalize=True,
        do_resize=True,
        reducing_gap=None,
        resample="lanczos",
        vae_latent_channels=4,
        vae_scale_factor=8
    )
    return processor.preprocess(image=images)

from torchvision.transforms import Compose, Resize, CenterCrop, ToTensor, Normalize
from torchvision.transforms import InterpolationMode
    
def _convert_image_to_rgb(image):
    return image.convert("RGB")


def _transform(n_px):
    return Compose([
        Resize(n_px, interpolation=InterpolationMode.BICUBIC),
        CenterCrop(n_px),
        _convert_image_to_rgb,
        ToTensor(),
        Normalize((0.48145466, 0.4578275, 0.40821073), (0.26862954, 0.26130258, 0.27577711)),
    ])


def clip_preprocess(images):
    return _transform(224)(images)
```
